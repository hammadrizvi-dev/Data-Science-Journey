## Why Data Visualization?

Data visualization converts numerical information into visual representations such as charts and graphs.

Visualization helps us:

- Identify patterns
- Understand relationships between variables
- Compare groups
- Detect unusual observations
- Understand data distribution
- Communicate analytical findings more effectively

A visualization should always answer an analytical question rather than being created without a specific purpose.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("D:/02_Courses_Learning/Data Science/Data-Science-Journey/Projects/Students_performance_analysis/Data/student_performance.csv")
```

### Does study time appear to be related to final marks?

```python
plt.scatter(df["study_hours"], df["final_marks"])

plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours vs Final Marks")

plt.show()
```

## Study Hours vs Final Marks

### Analytical Question

Does study time appear to be related to final marks?

### Visualization

A scatter plot is used to examine the relationship between study hours and final marks.

### Why a Scatter Plot?

A scatter plot is appropriate because both study hours and final marks are numerical variables. Each point represents one student.

### Interpretation

The scatter plot allows us to observe whether students with higher study hours generally tend to have higher or lower final marks.

### Important Note

An observed relationship between study hours and final marks does not necessarily mean that study hours cause higher final marks. Other factors may also influence student performance.

### Observation

The scatter plot shows a generally positive relationship between study hours and final marks. As study hours increase, final marks tend to increase as well.

However, the observations are relatively scattered rather than closely following a single pattern. This suggests that the relationship is not extremely strong.

A few observations also appear to differ from the overall pattern, indicating the presence of potentially unusual observations.

### Conclusion

In this dataset, students with higher study hours generally tend to have higher final marks. However, study hours alone do not fully explain variations in final marks, as students with similar study hours can have noticeably different final marks.

```python
df["study_hours"].corr(df["final_marks"])
```

### Correlation Analysis

To quantify the linear relationship between study hours and final marks, we calculate the Pearson correlation coefficient.

The correlation coefficient ranges from -1 to +1:

- A positive value indicates a positive linear relationship.
- A negative value indicates a negative linear relationship.
- A value close to zero indicates little or no linear relationship.

For this dataset, the correlation coefficient between study hours and final marks is approximately **0.48**.

### Interpretation

The result indicates a moderate positive linear relationship between study hours and final marks. This supports the pattern observed in the scatter plot, where final marks generally tend to increase as study hours increase.

However, correlation does not imply causation. Therefore, this result should not be interpreted as proof that increasing study hours directly causes higher final marks.

```python
correlation = df["study_hours"].corr(df["final_marks"])
correlation
```

```python
plt.scatter(df["attendance"],df["final_marks"])
plt.xlabel("Attendance")
plt.ylabel("Final Marks")
plt.title("Attendance vs Final Marks")
plt.show()
        
```

```python
attendance_correlation = df["attendance"].corr(df["final_marks"])
attendance_correlation
```

### Attendance vs Final Marks — Interpretation

The scatter plot shows a generally positive relationship between attendance and final marks. Students with higher attendance tend to have higher final marks, although the observations are relatively scattered.

Several observations do not follow the overall pattern. Some students with high attendance have relatively low final marks, while some students with lower attendance have relatively high final marks.

### Correlation Analysis

The Pearson correlation coefficient between attendance and final marks is approximately **0.39**.

This indicates a moderate positive linear relationship between attendance and final marks in this dataset.

### Comparison with Study Hours

The correlation between study hours and final marks is approximately **0.48**, which is higher than the attendance correlation of approximately **0.39**.

```python
plt.hist(df["final_marks"], bins=5)

plt.xlabel("Final Marks")
plt.ylabel("Students")
plt.title("Distribution of Final Marks")

plt.show()
```

## Distribution of Final Marks

### Analytical Question

How are the final marks distributed among the students?

### Visualization

A histogram is used to understand the distribution of final marks across the students.

### Why a Histogram?

A histogram is appropriate for examining the distribution of a numerical variable. It groups values into intervals called bins and shows the number of observations within each interval.

### Observation

The final marks are distributed across a range of approximately 50 to 76. A larger concentration of students can be observed around the middle-to-upper range of the marks.

### Conclusion

The histogram provides an overview of how final marks are distributed across the dataset and helps identify where the observations are concentrated.

```python
plt.hist(df["final_marks"], bins=15)

plt.xlabel("Final Marks")
plt.ylabel("Students")
plt.title("Distribution of Final Marks")

plt.show()
```

### Effect of Bin Size

The number of bins affects how the distribution of a numerical variable is visualized.

With fewer bins, such as 5 bins, the distribution appears smoother and provides a broader overview of where the observations are concentrated.

With more bins, such as 15 bins, the distribution provides more detail but may also appear more irregular because smaller intervals contain fewer observations.

Changing the number of bins does not change the underlying data. It only changes how the data is grouped and visualized.

Therefore, bin selection should be based on the dataset and the analytical purpose of the visualization.

```python
plt.scatter(df["previous_marks"], df["final_marks"])

plt.xlabel("Previous Marks")
plt.ylabel("Final Marks")
plt.title("Previous Marks vs Final Marks ")
plt.show
```

## Comparison of Variables

To compare the linear relationships between different variables and final marks, Pearson correlation coefficients were calculated.

| Variable | Correlation with Final Marks |
|---|---:|
| Study Hours | 0.48 |
| Previous Marks | 0.46 |
| Attendance | 0.39 |

### Interpretation

Among the variables analyzed, study hours show the strongest positive linear correlation with final marks, with a correlation coefficient of approximately 0.48.

Previous marks show a very similar positive correlation of approximately 0.46, while attendance has a comparatively lower correlation of approximately 0.39.

All three variables show positive relationships with final marks, but none of the correlations are close to +1. Therefore, final marks are not explained by any single variable alone.

### Important Limitation

Correlation measures linear association and does not establish causation. These results describe relationships observed within this dataset and should not be interpreted as proof that any of these variables directly causes higher final marks.
