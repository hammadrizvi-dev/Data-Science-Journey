# Groupby

```python
df.groupby("gender")
```

```python
(df.groupby("gender")
    ["final_marks"].mean())
```

```python
df.groupby("gender")[
    ["study_hours", "attendance", "final_marks"]
    ].mean()
```

```python
df.groupby("gender")[
    ["study_hours", "attendance", "final_marks"]
    ].agg(["mean", "min", "max"])
```

```python
df.groupby("gender")[
    ["final_marks"]
    ].agg(["mean", "min", "max"])
```

```python
df.groupby("gender").agg({
    "study_hours": "mean", "attendance" : "mean", "final_marks" : ["mean", "min", "max"] })
```

```python
df.groupby("gender")[
    ["final_marks", "previous_marks"]
    ].mean()
```
