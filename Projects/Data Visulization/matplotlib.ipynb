```python
import pandas as np
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.xlabel("Axis X")
plt.title("Hello")
plt.ylabel("axis Y")
```

```python
plt.subplot(2,2,2)
plt.plot(y**2,x**2)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,3)
```

### Object Oriented way

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.5,0.3,0.3])
axis2.plot(x,y)
```

### Types of Plot

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

### Saving of plot

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("Basicplot.png")
```
